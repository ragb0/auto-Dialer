```python
import pandas as pd
import os
import time
import subprocess
```

```python
# Load the Excel sheet with contact information
df = pd.read_excel('D:/contacts.xlsx')
df
```

```python
# Path to adb executable
adb_path = 'D:/platform-tools/adb'
```

```python
# Iterate over each contact and make a call
for index, row in df.iterrows():
    name = row['name']
    phone_number = str(row['number']).replace(" ", "")  # Remove spaces from the phone number

    # Show the name of the person being called
    print(f"Calling {name} at {phone_number}")

    # Ensure the number starts with '+'
    if not phone_number.startswith('+'):
        phone_number = '+' + phone_number

    # Construct the ADB command
    adb_command = [adb_path, 'shell', 'am', 'start', '-a', 'android.intent.action.CALL', '-d', f'tel:{phone_number}']

    # Print the command for debugging purposes
    print(f"Executing command: {' '.join(adb_command)}")

    # Execute the ADB command using subprocess
    result = subprocess.run(adb_command, capture_output=True, text=True)

    # Print the command result and output
    print(f"Command result code: {result.returncode}")
    print(f"Command output: {result.stdout}")
    print(f"Command error: {result.stderr}")

    # Check if the command executed successfully
    if result.returncode == 0:
        print(f"Call initiated to {phone_number}")
    else:
        print(f"Failed to initiate call to {phone_number}")

    # Wait for the call to be completed (adjust the time as needed)
    time.sleep(30)  # Delay between calls (in seconds)
```
